We are going to split this dataset in two separate data .

```python
# One-hot encode the 'gender' column (drop_first=True to avoid multicollinearity, optional for trees)
input_dataset = pd.get_dummies(input_dataset, columns=['gender'], drop_first=True)
print(input_dataset)
```

```python
output_dataset = music_df['genre']
print(output_dataset)
```

we need to split the data sets into two sets.
One for training and another for testing.The usual rule is to allocate 70% to 80% of the data for training and the other 20% to 30% for testing.

First we import the "train_test_split" function from the "sklearn.model_selection" module.
This function takes tree arguments the input data set "X" , the output data set "y", and a parameter to specify the size of the test dataset "test_size".
This function returns a tuple, so we can unpack this tuple to four variables.


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(input_dataset, output_dataset, test_size=0.2)
```

```python
from sklearn.tree import DecisionTreeClassifier

model = DecisionTreeClassifier()

# Now we have a model , we need to train it to learn patterns in the dataset. To do that we call the method ".fit()". This method takes two dataset, the input and the output.
model.fit(X_train, y_train)
```

Next we need to ask our model to make a prediction. To do that we call the ".predict()" method. This method takes a two dimensional array.
In this two dimensional array each element is an array, we pass a new input set.
In our case we are passing two inputs simultaneously.
**A 21 year old male and a 22 year old female. We put the result in a variable called predictions.**

**predictions = model.predict( [ [21, 1], [22, 0] ] )**

```python
predictions = model.predict(X_test)
print(predictions)
```

Now to calculate the accuracy we import **"accuracy_score()"** function from the **"sklearn.metrics"** module.
The **"accuracy_score()"** takes two methods,the output test dataset**"y_test" **and **our predictions** and returns an accuracy score between 0 to 1.
We store the result in the variable "model_score"

```python
from sklearn.metrics import accuracy_score
model_score = accuracy_score(y_test, predictions)
print(model_score)
```
